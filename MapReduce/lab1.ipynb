```python
#!/usr/bin/env python3
import sys

for line in sys.stdin:
    parts = [p.strip() for p in line.strip().split(",")]
    if len(parts) != 5:
        print(f"ERROR\t{line.strip()}|thieu_truong", file=sys.stderr)
        continue
    ma, cn, nv, gt, tt = parts
    try:
        gt = float(gt)
    except ValueError:
        print(f"ERROR\t{ma}|gia_tri_khong_phai_so", file=sys.stderr)
        continue
    if tt == "THANH_CONG" and gt >= 0:
        print(f"{nv}\t{gt}")
    else:
        print(f"REJECT\t{ma}|{tt}|{gt}", file=sys.stderr)  # để đối soát
```

```python
#!/usr/bin/env python3
import sys

cur, total = None, 0.0
for line in sys.stdin:
    key, val = line.rstrip("\n").split("\t")
    if key != cur:
        if cur is not None:
            print(f"{cur}\t{total:g}")
        cur, total = key, 0.0
    total += float(val)
if cur is not None:
    print(f"{cur}\t{total:g}")
```
